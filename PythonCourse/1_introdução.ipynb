```python
lanche = ('☕','🥛','🍞','🍰')
```

```python
print(lanche[3]) #sempre começa de 0 então o pão ficaria  ☕ = 0 ,🥛 = 1,🍞 = 2,🍰 = 3
```

```python
lanche[3]= '🍮' # o erro aqui , é por que não se poe alterar, POR SE TRATAR DE UMA TUPLA.
```

```python
print(lanche) #sem alteração
```

```python
lanche = ['☕','🥛','🍞','🍰'] #vamos tentar alterear quando se trata de uma lista.
```

```python
lanche[3]= '🍮' #aqui vai substituir a posição 3 por 🍮
```

```python
print(lanche)# aqui trocamos, por se tratar de LISTAS, listas podem ser alteradas no decorrer do código.
```

```python
lanche.append('🍪')# adicionamos como o apendice, mais uma coisa na nossa lista, nem preciso flar que se fosse tuples nao daria né?
```

```python
print(lanche)
```

```python
lanche.remove('🍞') # removemos um elmento.
```

```python
print(lanche) #easey pesy
```

```python
if '🍪' in lanche:
    lanche.remove("🍪")
# se um elemento estiver dentro de lanche, remova-o.
```

```python
print(lanche)
```
