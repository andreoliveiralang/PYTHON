d) Cria uma lista com idade

Resposta: Alternativa C
A função input() captura a entrada do usuário como uma string. Ao envolver o input() com int(), o Python converte esse texto em um número inteiro. Isso permite que o programa realize operações matemáticas com o valor recebido, o que não seria possível se ele permanecesse como texto.

Q8. Considere:
import numpy as np
media = np.mean([10, 20, 30])
print("Média = {:.2f}".format(media))
Qual será a saída?
a) Média = 20
b) Média = 20.0
c) Média = 20.00
d) Erro

```python
import numpy as np
media = np.mean([10, 20, 30])
print("Média = {:.2f}".format(media))
```

Resposta: Alternativa C
O Cálculo: O NumPy calcula a média de 10, 20 e 30, resultando em 20.0 (o NumPy geralmente trabalha com números decimais, ou floats).
