** Create a function that grabs the email website domain from a string in the form: **

    user@domain.com
    
**So for example, passing "user@domain.com" would return: domain.com**

```python
def pegarDomínio(email):                         # def = define = define uma nova função!
    return email.split('@')[1]                  # entre parenteses ali, é o que vai ser pedido para colocar, o INPUT.
                                                 # Return faz com que a operaão na frente dele seja guardado, seja salvo 
```

```python
pegarDomínio('andre_luyzz@hotmail.com')
```

** Create a basic function that returns True if the word 'dog' is contained in the input string. Don't worry about edge cases like a punctuation being attached to the word dog, but do account for capitalization. **

```python
def findDog(frase):
    fraseMinu = frase.lower() # transofrma a frase em minusculo, faznendo com que qualquer jeito que apessoa digitar, o python vai ler ela em minuscula.
    return 'dog' in fraseMinu # IN= O in sempre faz uma pergunta para o Python: "Isso aqui está contido naquilo ali?". O Python sempre responde com um valor booleano: True (Verdade) ou False (Falso).
```

```python
findDog('Is there a dog here?')
```

** Create a function that counts the number of times the word "dog" occurs in a string. Again ignore edge cases. **

```python
def countDog(frase):
    fraseMinu = frase.lower()
    return fraseMinu.count('dog') #ele vai pegar a frase que foi transofmrada em minusculo e contar a palavra dog.
```

```python
countDog('This dog runs faster than the other dog dude!')
```
