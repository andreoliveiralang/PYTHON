### Criando um dataframe a partir de um dicionário

```python
# dataframe = pd.DataFrame()
venda = {'data': ['15/02/2021', '16/02/2021'],
         'valor': [500, 300],                                     # isso aqui é oque criamos , ja esta tudo atribuido como vendas.
         'produto': ['feijao', 'arroz'],
         'qtde': [50, 70],
        }
vendasDF = pd.DataFrame(venda) #DF = DATA FRAME
```

### Importando arquivos e bases de dados

```python
#vamos importar de uma base de dados, no caso um arquivo excel
vendasDF = pd.read_excel("Vendas.xlsx")
display(vendasDF)
#rows = linhas  - coluns = colunas
```

```python
display(vendasDF.head(6))   # o metodo ❗HEAD❗ aqui mostra a cabeça da nossa tabela, o numero dentro é ate qual numero irá mostrar
display(vendasDF.tail(6))   # o metodo ❗TAIL❗ aqui vemos o cauda, ou seja o final.
                            # poderia ser print ao invez de display, mas o display mostra bem mais organizado. 
display(vendasDF.shape)     # o método ❗SHAPE❗, a gente vê quantas linhas tem e quantas colunas
display(vendasDF.describe())# o metodo ❗DESCRIBE❗faz um resumo, mosstra em média os dados, m´pedia de valores, unidades vendidas e tal.
```

### .loc, um método muito importante
    - Pegar 1 linha
    - Pegar linhas de acordo com alguma condição
    - Pegar linhas e colunas específicas
    - Pegar 1 valor específico

```python
#como pegar uma linha especifica?
display(vendasDF.loc[0]) #basicamente isso, e nas chaves vc adiciona a linha, pelo INDICE, que é a primeira coluna.
```

```python
display(vendasDF.loc[0:5]) # aqui eu poderia taambém pegar mais de uma linha, por exemplo ✨ usando os dois pontos ":" ✨ 
```

```python
display(vendasDF.loc[vendasDF["ID Loja"] == "Iguatemi Esplanada"]) # aqui entramos na coluna, ID LOJAS e filtramos todas as vezes que é IGUAL a 'iguatemi Esplanda
```

```python
display(vendasDF.loc[vendasDF["Produto"] == "Relógio"]) # filtramos aqui nma coluna produtos e tudo que fosse igual a rélogio ele filtrou.
```

```python
#pegar varias linhas e colunas usando o loc
vendasIguatemi = vendasDF.loc[vendasDF["ID Loja"] == "Iguatemi Esplanada",["ID Loja","Produto","Quantidade"]]  # aqui definimos que a variavel "vendasIguatemi" é a tabela filtrada para mostrar somente as vendas dessa loja.
display(vendasIguatemi)
```

```python
#PEGAR UM VALOR ESPECÍFICO.

#print(vendasDF[linha,coluna])
print(vendasDF.loc[46,"Produto"]) # tem que voltar relogio
```

### Adicionar 1 coluna

```python
#adicionar coluna nova, apartir de uma coluna que ja existe, usamos uma que existe para criar uma nova... exemplo:
vendasDF["Comissão"] = vendasDF["Valor Final"] * 0.05 # pega o valor fiunal da vendasdf e mulçtiplica para criar uma variavel comissão
display(vendasDF)
#veja que criou a tabela comissão apartir da coluna valor final!
```

```python
#agora vamos criar uma coluna totalmente do zero,
#vendasDF.loc[linha,coluna] = 0
vendasDF.loc[:,"imposto"] = 0 #quando adicionamos dois pontos aqui, quer dizer que o valor 0 vai ser em ✨todas as linhas✨.
display(vendasDF)
```

### Adicionar 1 linha
    - Linhas de um complemento da base de dados

```python
#como adicionar linha no nosso data frame
vendasDezembroDF = pd.read_excel("Vendas - Dez.xlsx") #importa arquivo de vendas dezembro para codigo.
display(vendasDezembroDF)

vendasDF = pd.concat([vendasDF, vendasDezembroDF]) # aqui adiciona como apendice os dados da variavel que criei , as vendas de dezembro
display(vendasDF)
```

### Excluir linhas e colunas

```python
vendasDF = vendasDF.drop("imposto",axis = 1) # eixo = 0 linha, eixo = 1 coluna
display(vendasDF)
```

### Valores Vazios
    - Deletar linhas/colunas vazias
    - Deletar linhas que possuem valores vazios
    - Preencher valores vazios (média e último valor)

```python
#metodo de exluir linhas e colunas vazios vazios NAN
vendasDF = vendasDF.dropna(how="all",axis=1) #aqui basicamente exlui a coluna toda, se tiver tudo vazio.

```

```python
#deletar linhas que possuem pelo menos 1 valor vazio
vendasDF = vendasDF.dropna() #sem inserir paramentro ja faz oque queremos.
```

```python
#preencer com a média da coluna
vendasDF['Comissão'] = vendasDF ["Comissão"].fillna(vendasDF["Comissão"].mean())

# acessa dataframe entrando na coluna comissão, onde  define essa coluna novamente fazendo com que dessa vez "fillna" ou seja ppreencha os valores vazios com  o valor da MÉDIA(mean) da comissao - FORMA MAIS USADA DE PREENCHER
display(vendasDF)
```

```python
#preencher com o ultimo valor
vendasDF = vendasDF.ffill()
display(vendasDF)
```

### Calcular Indicadores
    - Groupby
    - Value Counts

```python
transacoesLojas = vendasDF["ID Loja"].value_counts()
display(transacoesLojas)

#basicamente faz a contagem, no caso a quantidade de transições 
```

```python
#agrupa informações

#ele vai agrupar e somar, nesse exemplo , ele vbai agrupar todos itens chamados camisas e somar os valores.

# faturamentoProduto = vendasDF.groupby("coluna") 
faturamentoProduto = vendasDF.groupby("Produto").sum()

#pega o vendasDF, vai agrupar a coluna de produtos( aparecer um produto só em cada linha) e somar as colunas numéricas

display(faturamentoProduto) 
```
